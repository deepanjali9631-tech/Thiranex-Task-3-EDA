```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("Online_Retail.csv")

print("Dataset loaded successfully!")
df.head()
```

```python
missing_values = df.isnull().sum()

missing_percentage = (missing_values / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': missing_values,
    'Percentage': missing_percentage
})

missing_df.sort_values(by='Missing Values', ascending=False)
```

```python
plt.figure(figsize=(10, 5))

missing_percentage[missing_percentage > 0].sort_values(ascending=False).plot(
    kind='bar'
)

plt.title("Percentage of Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Missing Percentage")
plt.xticks(rotation=45)
plt.show()
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
duplicate_percentage = (df.duplicated().sum() / len(df)) * 100

print(f"Duplicate percentage: {duplicate_percentage:.2f}%")
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
df = df.dropna(subset=['CustomerID'])

print("Shape after removing missing CustomerID:", df.shape)
```

```python
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]

print("Shape after removing invalid transactions:", df.shape)
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

df['CustomerID'] = df['CustomerID'].astype(int)

df.info()
```

```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']

df.head()
```

```python
df[['Quantity', 'UnitPrice', 'Revenue']].describe()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(df['Quantity'], bins=50, kde=True)

plt.title("Distribution of Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(df['UnitPrice'], bins=50, kde=True)

plt.title("Distribution of Unit Price")
plt.xlabel("Unit Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(df['Revenue'], bins=50, kde=True)

plt.title("Distribution of Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(10, 4))

sns.boxplot(x=df['Quantity'])

plt.title("Box Plot of Quantity")
plt.xlabel("Quantity")
plt.show()
```

```python
plt.figure(figsize=(10, 4))

sns.boxplot(x=df['UnitPrice'])

plt.title("Box Plot of Unit Price")
plt.xlabel("Unit Price")
plt.show()
```

```python
top_products_quantity = (
    df.groupby('Description')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_quantity
```

```python
plt.figure(figsize=(10, 6))

top_products_quantity.sort_values().plot(kind='barh')

plt.title("Top 10 Products by Quantity Sold")
plt.xlabel("Quantity Sold")
plt.ylabel("Product")
plt.show()
```

```python
top_products_revenue = (
    df.groupby('Description')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products_revenue
```

```python
plt.figure(figsize=(10, 6))

top_products_revenue.sort_values().plot(kind='barh')

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product")
plt.show()
```

```python
top_countries = df['Country'].value_counts().head(10)

top_countries
```

```python
plt.figure(figsize=(10, 5))

top_countries.plot(kind='bar')

plt.title("Top 10 Countries by Number of Transactions")
plt.xlabel("Country")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.show()
```

```python
country_revenue = (
    df.groupby('Country')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

country_revenue
```

```python
plt.figure(figsize=(10, 6))

country_revenue.sort_values().plot(kind='barh')

plt.title("Top 10 Countries by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Country")
plt.show()
```

```python
customer_revenue = (
    df.groupby('CustomerID')['Revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

customer_revenue
```

```python
plt.figure(figsize=(10, 6))

customer_revenue.sort_values().plot(kind='barh')

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Customer ID")
plt.show()
```

```python
unique_customers = df['CustomerID'].nunique()

print("Number of unique customers:", unique_customers)
```

```python
unique_products = df['Description'].nunique()

print("Number of unique products:", unique_products)
```

```python
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Day'] = df['InvoiceDate'].dt.day
df['Hour'] = df['InvoiceDate'].dt.hour
df['DayOfWeek'] = df['InvoiceDate'].dt.day_name()
```

```python
monthly_revenue = (
    df.groupby(['Year', 'Month'])['Revenue']
    .sum()
    .reset_index()
)

monthly_revenue['YearMonth'] = (
    monthly_revenue['Year'].astype(str)
    + "-"
    + monthly_revenue['Month'].astype(str).str.zfill(2)
)

monthly_revenue
```

```python
plt.figure(figsize=(12, 5))

sns.lineplot(
    data=monthly_revenue,
    x='YearMonth',
    y='Revenue',
    marker='o'
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
day_order = [
    'Monday',
    'Tuesday',
    'Wednesday',
    'Thursday',
    'Friday',
    'Saturday',
    'Sunday'
]

day_revenue = (
    df.groupby('DayOfWeek')['Revenue']
    .sum()
    .reindex(day_order)
)

day_revenue
```

```python
plt.figure(figsize=(10, 5))

day_revenue.plot(kind='bar')

plt.title("Revenue by Day of Week")
plt.xlabel("Day")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
hourly_transactions = df['Hour'].value_counts().sort_index()

plt.figure(figsize=(10, 5))

hourly_transactions.plot(kind='line', marker='o')

plt.title("Transactions by Hour")
plt.xlabel("Hour")
plt.ylabel("Number of Transactions")
plt.xticks(range(0, 24))
plt.show()
```

```python
numeric_columns = [
    'Quantity',
    'UnitPrice',
    'Revenue'
]

correlation = df[numeric_columns].corr()

correlation
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Quantity',
    y='Revenue',
    alpha=0.5
)

plt.title("Quantity vs Revenue")
plt.xlabel("Quantity")
plt.ylabel("Revenue")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='UnitPrice',
    y='Revenue',
    alpha=0.5
)

plt.title("Unit Price vs Revenue")
plt.xlabel("Unit Price")
plt.ylabel("Revenue")
plt.show()
```

```python
invoice_revenue = df.groupby('InvoiceNo')['Revenue'].sum()

average_order_value = invoice_revenue.mean()

print(f"Average Order Value: {average_order_value:.2f}")
```

important patterns in customer purchasing behavior, product performance,
country-level sales, and revenue trends.
